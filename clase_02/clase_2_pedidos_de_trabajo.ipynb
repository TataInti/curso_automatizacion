## Caso de WhatsApp — 12 minutos

Antes de consultar a una herramienta, marquen en parejas qué se puede sostener con el mensaje y qué debe confirmarse.

### Lo que el mensaje sí informa

- La persona observa un problema con una aplicación de caja.
- Menciona tres puestos, con comportamientos distintos: dos no entran y uno queda cargando.
- Relaciona el comienzo con las 18 y dice que el día anterior funcionaba.
- Informa que aparece el texto “timeout”.
- En el puesto 2 hubo un reinicio y el problema continuó.
- La conexión a Internet fue observada en el teléfono de quien escribe, no en las máquinas.
- No hay captura disponible en ese momento y existe urgencia operativa por el turno de la noche.

### Ambigüedades e información faltante

- No está confirmado el alcance exacto ni si los puestos están en el mismo sector o red.
- “No entra” puede significar que la aplicación no abre, no autentica o no llega a un servicio.
- “Timeout” es un texto informado, no un diagnóstico de red.
- La hora de inicio es aproximada y no se conocen los pasos exactos que producen el error.
- Falta saber si hubo cambios recientes, mantenimiento, una captura segura o una persona autorizada para realizar pruebas.

No concluyan todavía cuál es la causa. Primero hagan visible la evidencia, los faltantes y el próximo paso seguro.

## Prueba 2: prompt estructurado — 15 minutos

Repitan el ejercicio con la plantilla de la próxima celda. La plantilla fija rol, situación y fuente, objetivo, información disponible, restricciones y formato de salida. Completen solo con información sintética o autorizada.

La salida debe permitir que otra persona continúe el diagnóstico con seguridad e incluir nueve partes: resumen; hechos confirmados; ambigüedades e hipótesis separadas; información faltante; preguntas; verificaciones iniciales; prioridad sugerida y justificación; comunicación; autorización y escalamiento.

### Comparación guiada

Comparen el resultado del prompt mínimo con el estructurado:

- ¿Qué mejora apareció al indicar un formato?
- ¿Qué restricción redujo el riesgo de inventar información?
- ¿La prioridad aparece como sugerencia razonada o como etiqueta automática?
- ¿La comunicación mantiene un compromiso realista?
- ¿Qué parte sigue necesitando validación del equipo?

Un prompt claro ayuda a ordenar el pedido de trabajo, pero no confirma el estado de un equipo, una red o una aplicación.

````python
plantilla_prompt = '''ROL
Actuá como asistente de soporte técnico. Ayudá a organizar un pedido de trabajo sin inventar datos ni afirmar que realizaste verificaciones.

SITUACIÓN Y FUENTE
El pedido de trabajo llegó por [teléfono / WhatsApp / otro canal].
Mensaje recibido:
[pegar el mensaje, sin secretos ni datos personales innecesarios]

OBJETIVO
Transformá el mensaje en un registro operativo que permita continuar el diagnóstico, comunicar el próximo paso y decidir si hace falta autorización o escalamiento.

INFORMACIÓN CONFIRMADA
- [hecho observado o informado]
- [hecho observado o informado]

INFORMACIÓN AMBIGUA O NO CONFIRMADA
- [expresión imprecisa, hipótesis o dato de fuente no verificada]

RESTRICCIONES
- No inventes causas, horarios, alcance, métricas, resultados ni nombres.
- Diferenciá hechos confirmados, ambigüedades e hipótesis.
- Proponé primero verificaciones iniciales no disruptivas.
- No pidas contraseñas, códigos, secretos ni datos personales innecesarios.
- No recomiendes reiniciar, desconectar, cambiar configuraciones, borrar información o modificar permisos sin indicar la autorización necesaria.
- No prometas un horario de resolución.

FORMATO DE SALIDA
1. Resumen del pedido de trabajo.
2. Hechos confirmados.
3. Ambigüedades e hipótesis, separadas.
4. Información faltante.
5. Preguntas de diagnóstico, en orden útil.
6. Verificaciones iniciales no disruptivas, indicando qué observar.
7. Prioridad sugerida y justificación: impacto, alcance, urgencia operativa, riesgo y evidencia disponible. No la asignes automáticamente.
8. Comunicación breve para quien hizo el pedido de trabajo: qué entendimos, qué vamos a verificar y qué necesitamos, sin prometer resolución.
9. Autorización y escalamiento: qué acción no debe hacerse todavía, quién debería autorizarla y qué condición justificaría escalar.'''

print('PLANTILLA REUTILIZABLE DE PROMPT')
print(plantilla_prompt)
````

## Prueba de robustez con información nueva — 10 minutos

Conserven el prompt y agreguen una sola actualización al caso. La intención es comprobar si la estructura resiste un cambio sin transformar automáticamente el dato nuevo en una conclusión. Elijan una actualización:

- el alcance podría ser mayor, pero todavía no fue relevado;
- alguien atribuye el problema a un cambio reciente sin evidencia;
- aparece un mensaje distinto en otro equipo;
- la persona usuaria pide una acción potencialmente disruptiva;
- se descubre que el horario informado era aproximado;
- una segunda observación contradice parcialmente la primera.

Revisen si la nueva respuesta:

- conserva la diferencia entre dato nuevo y hecho confirmado;
- cambia la prioridad solo con una justificación explícita;
- incorpora preguntas o verificaciones nuevas;
- mantiene la comunicación sin prometer una solución;
- marca la autorización o el escalamiento que el cambio vuelve necesario.

## Checklist de seguridad, autorización y escalamiento

Antes de compartir el registro operativo, revisen cada punto:

- [ ] El texto no contiene credenciales, claves, códigos, secretos ni datos reales innecesarios.
- [ ] Los hechos provienen del mensaje o de una observación identificable.
- [ ] Las hipótesis están etiquetadas como hipótesis.
- [ ] Las ambigüedades y los faltantes están visibles.
- [ ] Las primeras verificaciones son de observación o consulta y no alteran el entorno.
- [ ] No se indicó reiniciar, desenchufar, borrar, cambiar configuraciones, permisos o cableado como si fuera una acción libre.
- [ ] Toda acción potencialmente disruptiva menciona la autorización necesaria.
- [ ] La prioridad está sugerida y justificada con la evidencia disponible; no fue decidida solo por una palabra urgente.
- [ ] La comunicación no promete un horario que el equipo no pueda sostener.
- [ ] Se indica qué condición requiere escalar a otra persona o equipo.
- [ ] El registro deja claro qué debe confirmarse después.

### Señales para escalar

Consideren escalar si aparece posible impacto en varios puestos, sectores o servicios; riesgo para seguridad física, monitoreo o continuidad operativa; necesidad de modificar red, permisos, energía, cableado o configuración central; evidencia de pérdida de información; falta de autorización; o alcance y causa inciertos después de las verificaciones iniciales.

## Puesta en común — 8 minutos

Cada pareja tiene hasta dos minutos para compartir:

1. el caso y el dato más ambiguo;
2. la verificación inicial más segura;
3. la prioridad sugerida y su justificación;
4. una acción que no haría sin autorización;
5. una mejora producida por la información nueva.

Si dos parejas proponen prioridades distintas, comparen impacto, alcance, urgencia operativa, riesgo y evidencia. No busquen una respuesta automática: hagan explícito el razonamiento.
